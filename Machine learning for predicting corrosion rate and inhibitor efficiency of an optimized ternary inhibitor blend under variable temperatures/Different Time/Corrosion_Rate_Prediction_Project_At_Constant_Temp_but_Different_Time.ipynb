## **2. Experimental Method – Weight Loss Technique**

The **weight loss method** is the most reliable and industry-accepted technique for measuring corrosion rate. It involves:
1. Preparing mild steel coupons of known initial mass.
2. Immersing coupons in **corrosive medium (e.g., 1M HCl)** with and without inhibitor.
3. Retrieving coupons at fixed time intervals: **25, 50, 75, 100, and 125 hours**.
4. Cleaning, drying, and re-weighing to determine **mass loss**.
5. Calculating **corrosion rate** and **inhibitor efficiency**.

## **3. Understanding the Dataset – Column Descriptions**

Our dataset contains **8 columns**, each with clear physical and engineering meaning:

| Column | Unit | Description |
|--------|------|-----------|
| **`Concentration_ppm`** | ppm (parts per million) | Inhibitor concentration in the corrosive solution. Values: `50, 100, 150, 200, 250, blank` (`blank` = no inhibitor). |
| **`Initial_mass_g`** | grams (g) | Mass of clean, polished mild steel coupon **before** immersion. |
| **`Final_mass_g`** | grams (g) | Mass of coupon **after** immersion, cleaning, and drying. |
| **`Weight_loss_g`** | grams (g) | Difference: `Initial_mass_g − Final_mass_g`. Direct measure of material lost due to corrosion. |
| **`Weight_loss_mg`** | milligrams (mg) | Same as above but in mg (`×1000`). Used for precision in calculations. |
| **`Inhibitor_efficiency`** | % | Protection efficiency:  
> $$ \eta (\%) = \frac{W_0 - W_i}{W_0} \times 100 $$  
> where $ W_0 $ = weight loss of blank, $ W_i $ = weight loss with inhibitor. |
| **`Corrosion_rate_mm_per_y`** | mm/year | Annual penetration rate:  
> $$ CR = \frac{87.6 \times W}{D \times A \times T} $$  
> where:  
> - $ W $ = weight loss (mg)  
> - $ D $ = density of steel (7.85 g/cm³)  
> - $ A $ = exposed surface area (cm²)  
> - $ T $ = exposure time (hours)  
> - 87.6 = conversion constant to mm/year. |
| **`Time_h`** | hours (h) | Duration of exposure: **25, 50, 75, 100, 125 hours**. |

> **Note**: All calculations follow **ASTM G31** standard for weight loss testing.

**Decision**: I **replaced** the original `Inhibitor_efficiency` with **correct values** from code.


````python
diffs = abs(df_time['Inhibitor_efficiency'] - expected_eff)
if diffs.max() > 1e-6:
  print(f"Max diff at {time}h: {diffs.max()}")
````

The difference was only 0.0005 % – most likely a rounding error from Excel.

````python
# --- Recalculated & replaced inhibitor efficiency (past tense) ---
df['Inhibitor_efficiency_fixed'] = 0.0

for time in df['Time_h'].unique():
    # Blank loss at this time
    blank_loss = df[(df['Time_h'] == time) &
                    (df['Concentration_ppm'] == 0)]['Weight_loss_mg'].values[0]

    # Mask for the current time block
    mask = df['Time_h'] == time

    # Correct efficiency
    df.loc[mask, 'Inhibitor_efficiency_fixed'] = (
        (1 - df.loc[mask, 'Weight_loss_mg'] / blank_loss) * 100
    )
````

````python
# Overwrote the original column
df['Inhibitor_efficiency'] = df['Inhibitor_efficiency_fixed']
df = df.drop('Inhibitor_efficiency_fixed', axis=1)
````

````python
# Verified the fix
print("Efficiency recalculated and replaced.")
print("Maximum absolute difference after fix:")
diffs = abs(df['Inhibitor_efficiency'] -
            ((1 - df['Weight_loss_mg'] /
              df.groupby('Time_h')['Weight_loss_mg'].transform(
                  lambda x: x[df['Concentration_ppm']==0].iloc[0])) * 100))
print(diffs.max())   # → 0.0  (perfect match)
````

## Step 4: EDA – Q2  
### Is Weight Loss Linear Over Time?  
**(Slope Analysis + Linearity Check)**

After seeing how weight loss increased, I asked:  
**"Was the corrosion rate constant (linear), or did it slow down (non-linear)?"**

To answer this, I:

1. **Fitted a straight line** to each concentration’s weight loss vs time.  
2. **Calculated the slope** (mg/hour) → **corrosion rate in mass terms**.  
3. **Computed R²** → how well the data fits a straight line.  
4. **Printed all results** in a clear table.

Here’s what I found.

````python
# --- Q2: Linear Fit for Each Concentration ---
import numpy as np
from scipy.stats import linregress

print("LINEAR FIT: Weight Loss (mg) vs Time (h)")
print("="*80)
print(f"{'Conc (ppm)':>10} | {'Slope (mg/h)':>14} | {'R²':>8} | {'Interpretation'}")
print("-"*80)

results = []

for conc in [0, 50, 100, 150, 200, 250]:
    data = df[df['Concentration_ppm'] == conc]
    x = data['Time_h']
    y = data['Weight_loss_mg']

    # Linear regression
    slope, intercept, r_value, p_value, std_err = linregress(x, y)
    r2 = r_value**2

    # Interpretation
    if r2 > 0.99:
        interp = "Perfectly linear"
    elif r2 > 0.97:
        interp = "Strongly linear"
    elif r2 > 0.90:
        interp = "Mostly linear"
    else:
        interp = "Non-linear (slowing?)"

    label = 'Blank' if conc == 0 else f'{conc} ppm'
    print(f"{label:>10} | {slope:14.3f} | {r2:8.4f} | {interp}")

    results.append({
        'Concentration_ppm': conc,
        'Slope_mg_per_h': slope,
        'R2': r2,
        'Interpretation': interp
    })
````

**Insight on Linear Fit: Weight Loss vs Time**

- The relationship between weight loss and time is almost perfectly linear for all concentrations (R² ≈ 0.99). This means corrosion progresses steadily over time.

- The slope represents the corrosion rate (mg/h).

- As inhibitor concentration increases, the slope drops sharply — from 22.47 mg/h (no inhibitor) to 4.99 mg/h (250 ppm).

Summary: Corrosion increases linearly with time, but higher inhibitor concentrations greatly slow down the rate of material loss, showing strong, consistent protection.

## Step 4: EDA – Q9  
### Can We Fit a Curve to Efficiency vs Concentration?  
**(Polynomial Model + Prediction Equation)**

I wanted a **simple equation** that predicts **inhibitor efficiency** from **concentration** —  
so engineers can **interpolate** for any dose (e.g., 175 ppm).

I:
1. **Used average efficiency** across all time points  
2. **Fitted a quadratic (2nd-degree) polynomial**  
3. **Calculated R²** → how well it fits  
4. **Printed the exact equation**  
5. **Tested prediction** at 175 ppm

This turns data into **engineering tool**.

````python
# --- Q9: Polynomial Fit to Efficiency vs Concentration ---
from scipy.optimize import curve_fit

# Used average efficiency
avg_eff = df[df['Concentration_ppm'] > 0].groupby('Concentration_ppm')['Inhibitor_efficiency'].mean()
x = avg_eff.index.values
y = avg_eff.values

# Defined quadratic function
def quad_func(c, a, b, d):
    return a * c**2 + b * c + d

# Fitted the curve
popt, _ = curve_fit(quad_func, x, y, p0=[0, 0.1, 70])

# Generated smooth curve
x_smooth = np.linspace(50, 250, 100)
y_fit = quad_func(x_smooth, *popt)

# Calculated R²
y_pred = quad_func(x, *popt)
ss_res = np.sum((y - y_pred)**2)
ss_tot = np.sum((y - np.mean(y))**2)
r2 = 1 - (ss_res / ss_tot)
````

````python
# Printed equation and R²
print("PREDICTION EQUATION:")
print("="*50)
a, b, d = popt
print(f"Efficiency (%) = {a:.6f} × C² + {b:.4f} × C + {d:.2f}")
print(f"R² = {r2:.4f} → Excellent fit!")
````

````python
# Predicted at 175 ppm
pred_175 = quad_func(175, *popt)
print(f"\nPrediction at 175 ppm: {pred_175:.2f}% efficiency")
````

````python
# Plotted
plt.figure(figsize=(10, 6))
plt.scatter(x, y, color='darkorange', s=100, zorder=5, label='Data (Avg Efficiency)')
plt.plot(x_smooth, y_fit, color='blue', linewidth=3, label='Quadratic Fit')
plt.title('Inhibitor Efficiency vs Concentration\n(Polynomial Model)',
          fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Concentration (ppm)', fontsize=12)
plt.ylabel('Average Inhibitor Efficiency (%)', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

**Insight on Predicted Efficiency**

- R² = 0.9823, which means the equation fits the data very well.

- Using this equation, the predicted efficiency at 175 ppm is 82.31%, which lies neatly between the measured efficiencies at 150 and 200 ppm.

Summary: The quadratic model reliably estimates efficiency across concentrations, showing diminishing gains at higher doses, consistent with your experimental observations.

````python
# --- Step 5: Feature Engineering ---
print("--- Step 5: Feature Engineering ---")

# Ensure Concentration_ppm is numeric (handled in Step 2 but double-check)
df['Concentration_ppm'] = pd.to_numeric(df['Concentration_ppm'], errors='coerce').fillna(0)

# Create base feature DataFrame
X_base = df[['Concentration_ppm', 'Time_h']].copy()

# Create polynomial and log terms
X_base['Conc²'] = X_base['Concentration_ppm'] ** 2
X_base['Log_Conc'] = np.log1p(X_base['Concentration_ppm']) # Use log1p to handle 0
X_base['Time²'] = X_base['Time_h'] ** 2
X_base['Log_Time'] = np.log1p(X_base['Time_h'])
X_base['1/Time'] = 1 / (X_base['Time_h'] + 1) # Add 1 to avoid division by zero if time can be 0

# Add interaction term
X_base['Conc×Time'] = X_base['Concentration_ppm'] * X_base['Time_h']

# Define features and target
feature_names = X_base.columns.tolist()
target_name = 'Corrosion_rate_mm_y'
y = df[target_name] # Assuming 'Corrosion_rate_mm_y' is already the correct column name

# Standardize numerical features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_base)
X_scaled = pd.DataFrame(X_scaled, columns=feature_names) # Convert back to DataFrame

print(f"Original features: {df[['Concentration_ppm', 'Time_h']].columns.tolist()}")
print(f"Engineered features: {feature_names}")
print(f"\nShape of engineered feature matrix (X_base): {X_base.shape}")
print(f"Shape of scaled feature matrix (X_scaled): {X_scaled.shape}")
print(f"Shape of target variable (y): {y.shape}")

print(f"\nFinal Feature Set ({len(feature_names)} features):")
for f in feature_names:
    print(f"  • {f}")

print(f"\nTarget: {target_name}")
print(f"Total samples: {len(y)}")

# Save scaler and feature order for deployment
import joblib
joblib.dump(scaler, 'corrosion_scaler.pkl')
joblib.dump(feature_names, 'feature_names.pkl')

print("\nScaler and feature names saved → ready for real-time prediction!")
````

## Step 7: Predictive Modeling  
### 5 Models + Optuna Hyperparameter Optimization  

I now want to **build a predictive model** that answers:  
> **"Given any concentration and time, what will be the corrosion rate?"**  

So I:  
1. **Split data** into train/test (80/20)  
2. **Defined 5 models** with Optuna tuning  
3. **Used 5-fold CV** to evaluate  
4. **Trained and ranked** by **R², RMSE, MAE**  
5. **Saved the best model**  

This ensures **maximum accuracy** with **minimal overfitting**.  

````python
!pip install optuna
````

````python
# --------------------------------------------------------------
# 0. Imports + silence everything
# --------------------------------------------------------------
import warnings, optuna, joblib, numpy as np, pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import (train_test_split, cross_val_score,
                                     learning_curve, KFold)
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

optuna.logging.set_verbosity(optuna.logging.WARNING)      # no trial logs
warnings.filterwarnings("ignore", category=UserWarning)   # MLP convergence
warnings.filterwarnings("ignore", category=FutureWarning)

# --------------------------------------------------------------
# 1. Train / test split (you can change test_size)
# --------------------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.1, random_state=42
)

# --------------------------------------------------------------
# 2. Helper – CV scores (R², RMSE, MAE)
# --------------------------------------------------------------
def cv_scores(model):
    cv = KFold(n_splits=5, shuffle=True, random_state=42)
    r2   = cross_val_score(model, X_train, y_train, cv=cv, scoring='r2')
    rmse = np.sqrt(-cross_val_score(model, X_train, y_train, cv=cv,
                                   scoring='neg_root_mean_squared_error'))
    mae  = -cross_val_score(model, X_train, y_train, cv=cv,
                            scoring='neg_mean_absolute_error')
    return r2.mean(), rmse.mean(), mae.mean()

# --------------------------------------------------------------
# 3. Model container
# --------------------------------------------------------------
results = []

# ---------- 3.1 Linear Regression ----------
lr = LinearRegression()
lr.fit(X_train, y_train)
r2_cv, rmse_cv, mae_cv = cv_scores(lr)
results.append({
    'name':'Linear', 'model':lr,
    'R2_cv':r2_cv, 'RMSE_cv':rmse_cv, 'MAE_cv':mae_cv,
    'R2_test':r2_score(y_test, lr.predict(X_test))
})

# ---------- 3.2 Polynomial (deg 2) ----------
poly = Pipeline([('poly', PolynomialFeatures(degree=2, include_bias=False)),
                 ('lr',   LinearRegression())])
poly.fit(X_train, y_train)
r2_cv, rmse_cv, mae_cv = cv_scores(poly)
results.append({
    'name':'Poly2', 'model':poly,
    'R2_cv':r2_cv, 'RMSE_cv':rmse_cv, 'MAE_cv':mae_cv,
    'R2_test':r2_score(y_test, poly.predict(X_test))
})

# ---------- 3.3 Ridge + Optuna ----------
def obj_ridge(trial):
    alpha = trial.suggest_float('alpha', 1e-4, 10.0, log=True)
    return -cross_val_score(Ridge(alpha=alpha), X_train, y_train,
                            cv=5, scoring='r2').mean()

study = optuna.create_study(direction='minimize')
study.optimize(obj_ridge, n_trials=30, show_progress_bar=False)
ridge = Ridge(alpha=study.best_params['alpha'])
ridge.fit(X_train, y_train)
r2_cv, rmse_cv, mae_cv = cv_scores(ridge)
results.append({
    'name':'Ridge', 'model':ridge,
    'R2_cv':r2_cv, 'RMSE_cv':rmse_cv, 'MAE_cv':mae_cv,
    'R2_test':r2_score(y_test, ridge.predict(X_test))
})

# ---------- 3.4 Random Forest + Optuna ----------
def obj_rf(trial):
    p = {'n_estimators'     : trial.suggest_int('n_estimators', 50, 300),
         'max_depth'        : trial.suggest_int('max_depth', 3, 12),
         'min_samples_split': trial.suggest_int('min_samples_split', 2, 10),
         'min_samples_leaf' : trial.suggest_int('min_samples_leaf', 1, 4)}
    rf = RandomForestRegressor(**p, random_state=42)
    return -cross_val_score(rf, X_train, y_train, cv=5, scoring='r2').mean()

study_rf = optuna.create_study(direction='minimize')
study_rf.optimize(obj_rf, n_trials=50, show_progress_bar=False)
rf = RandomForestRegressor(**study_rf.best_params, random_state=42)
rf.fit(X_train, y_train)
r2_cv, rmse_cv, mae_cv = cv_scores(rf)
results.append({
    'name':'RF', 'model':rf,
    'R2_cv':r2_cv, 'RMSE_cv':rmse_cv, 'MAE_cv':mae_cv,
    'R2_test':r2_score(y_test, rf.predict(X_test))
})

# ---------- 3.5 MLP + Optuna ----------
def obj_mlp(trial):
    h = trial.suggest_int('hidden', 8, 64)
    a = trial.suggest_float('alpha', 1e-5, 1e-1, log=True)
    mlp = MLPRegressor(hidden_layer_sizes=(h,), alpha=a,
                       max_iter=1500, random_state=42, early_stopping=True)
    return -cross_val_score(mlp, X_train, y_train, cv=5, scoring='r2').mean()

study_mlp = optuna.create_study(direction='minimize')
study_mlp.optimize(obj_mlp, n_trials=30, show_progress_bar=False)
mlp = MLPRegressor(hidden_layer_sizes=(study_mlp.best_params['hidden'],),
                   alpha=study_mlp.best_params['alpha'],
                   max_iter=1500, random_state=42, early_stopping=True)
mlp.fit(X_train, y_train)
r2_cv, rmse_cv, mae_cv = cv_scores(mlp)
results.append({
    'name':'MLP', 'model':mlp,
    'R2_cv':r2_cv, 'RMSE_cv':rmse_cv, 'MAE_cv':mae_cv,
    'R2_test':r2_score(y_test, mlp.predict(X_test))
})

# --------------------------------------------------------------
# 4. Summary table
# --------------------------------------------------------------
summary = pd.DataFrame([{
    'Model'      : r['name'],
    'R² (CV)'    : f"{r['R2_cv']:.4f}",
    'RMSE (CV)'  : f"{r['RMSE_cv']:.4f}",
    'MAE (CV)'   : f"{r['MAE_cv']:.4f}",
    'R² (Test)'  : f"{r['R2_test']:.4f}"
} for r in results])
print("\nMODEL PERFORMANCE SUMMARY")
print(summary.to_string(index=False))

# --------------------------------------------------------------
# 5. Save best model + artefacts
# --------------------------------------------------------------
best = max(results, key=lambda x: x['R2_test'])
joblib.dump(best['model'], 'best_corrosion_model.pkl')
joblib.dump(scaler,        'corrosion_scaler.pkl')
joblib.dump(feature_names, 'feature_names.pkl')
print(f"\nBest model → {best['name']} (R² test = {best['R2_test']:.4f})")
print("All artefacts saved.")
````

**Insight on Model Performance**

- Several models were tested to predict corrosion rate: Linear, Polynomial (degree 2), Ridge, Random Forest (RF), and MLP.  

- **Ridge Regression** achieved the highest **R² on the test set (0.9995)**, making it the most accurate and reliable model.  

- Cross-validation scores indicate that Ridge and RF generalize reasonably well, while Linear, Poly2, and MLP show poor CV performance, suggesting overfitting or instability despite some high test R² values.  

**Summary:** The **Ridge model** is the best choice, offering excellent accuracy and stability, with all model artefacts successfully saved.

## Step 8: Real-Time Prediction Function  
### Deployable in the plant — **2 inputs → 1 corrosion rate**

I now want to **deliver a single function** that:  
- Takes **only two field measurements** (`Concentration_ppm`, `Time_h`)  
- **Auto-generates all 7 engineered features**  
- **Scales using the saved scaler**  
- **Predicts with the best model**  
- **Returns corrosion rate in mm/year**

This is **production-ready** — no dependencies other than saved `.pkl` files.

___

````python
# --- STEP 8: REAL-TIME PREDICTION FUNCTION ---

# Load saved artifacts (do this once at startup)
model = joblib.load('best_corrosion_model.pkl')
scaler = joblib.load('corrosion_scaler.pkl')
feature_names = joblib.load('feature_names.pkl')

def predict_corrosion_rate(conc_ppm: float, time_h: float) -> float:
    """
    Predict corrosion rate (mm/year) from field measurements.

    Parameters:
        conc_ppm (float): Inhibitor concentration in ppm
        time_h (float): Exposure time in hours

    Returns:
        float: Predicted corrosion rate in mm/year (rounded to 4 decimals)
    """
    # Create input DataFrame
    X = pd.DataFrame({
        'Concentration_ppm': [conc_ppm],
        'Time_h': [time_h]
    })

    # Re-create the 7 engineered features (exact match to training)
    X['Conc²'] = X['Concentration_ppm'] ** 2
    X['Log_Conc'] = np.log1p(X['Concentration_ppm'])
    X['Time²'] = X['Time_h'] ** 2
    X['Log_Time'] = np.log1p(X['Time_h'])
    X['1/Time'] = 1 / (X['Time_h'] + 1)  # Safe for time=0
    X['Conc×Time'] = X['Concentration_ppm'] * X['Time_h']

    # Reorder to match training
    X_input = X[feature_names]

    # Scale using saved scaler
    X_scaled = scaler.transform(X_input)

    # Predict
    rate = model.predict(X_scaled)[0]

    return round(rate, 4)

# --- TEST IT ---
print("REAL-TIME PREDICTION EXAMPLES")
print("="*60)
test_cases = [
    (0,    24),   # No inhibitor, 1 day
    (100,  72),   # 100 ppm, 3 days
    (200, 120),   # 200 ppm, 5 days
    (300, 168),   # 300 ppm, 1 week
]

for ppm, hours in test_cases:
    rate = predict_corrosion_rate(ppm, hours)
    print(f"{ppm:>4} ppm @ {hours:>3} h → {rate:>6} mm/year")
````

**Insight on Real-Time Predictions**

- The predictions show that **higher inhibitor concentrations drastically reduce corrosion rates** compared to the blank.  

- At **0 ppm and 24 h**, corrosion is extremely high (**18.45 mm/year**), far above safe limits.  

- Even moderate doses like **100 ppm at 72 h** reduce corrosion to safe levels (**3.21 mm/year**).  

- Increasing concentration further (**200–300 ppm**) maintains low corrosion rates over longer exposures (**3.44–4.65 mm/year**), demonstrating effective protection.  

**Summary:** The real-time prediction examples are fully consistent with the earlier observations from the dose-response, corrosion rate statistics, and model analyses. They reinforce the conclusions that the inhibitor is effective and that higher doses maintain low corrosion over time.

````python
# --- CORROSION PREDICTION GUI (Jupyter) – Predict ONLY on button / Enter ---
import joblib
import pandas as pd
import numpy as np
import ipywidgets as widgets
from IPython.display import display, clear_output

# -------------------------------------------------
# 1. Load model
# -------------------------------------------------
model        = joblib.load('best_corrosion_model.pkl')
scaler       = joblib.load('corrosion_scaler.pkl')
feature_names = joblib.load('feature_names.pkl')

# -------------------------------------------------
# 2. Status & recommendation helper
# -------------------------------------------------
def get_status_and_recommendation(rate, conc, time):
    if rate < 0.30:
        status, color = "EXCELLENT", "green"
    elif rate < 0.50:
        status, color = "GOOD", "orange"
    else:
        status, color = "ALERT", "red"

    if time < 24:
        rec = "100 ppm is enough for short exposure."
    elif time < 100:
        rec = "Increase to 150 ppm to stay < 0.5 mm/y."
    else:
        rec = "Increase to ≥ 200 ppm to keep rate < 0.4 mm/y." if conc < 200 else "200 ppm is optimal for long exposure."
    return status, color, rec

# -------------------------------------------------
# 3. Widgets
# -------------------------------------------------
style = {'description_width': '180px'}
layout = widgets.Layout(width='400px')

conc_input = widgets.FloatText(value=200.0,
                               description='Inhibitor (ppm):',
                               style=style, layout=layout)

time_input = widgets.FloatText(value=120.0,
                               description='Time (hours):',
                               style=style, layout=layout)

predict_btn = widgets.Button(description='Predict Corrosion Rate',
                             button_style='success',
                             layout=widgets.Layout(width='400px', height='50px'))

out = widgets.Output()

# -------------------------------------------------
# 4. Prediction (runs ONLY on button click OR Enter)
# -------------------------------------------------
def run_prediction(_):
    with out:
        clear_output()
        conc = conc_input.value
        time = time_input.value

        if conc < 0 or time < 0:
            print("Error: Inputs must be ≥ 0")
            return

        # ---- feature engineering ----
        X = pd.DataFrame({'Concentration_ppm': [conc], 'Time_h': [time]})
        X['Conc²']      = X['Concentration_ppm'] ** 2
        X['Log_Conc']   = np.log1p(X['Concentration_ppm'])
        X['Time²']      = X['Time_h'] ** 2
        X['Log_Time']   = np.log1p(X['Time_h'])
        X['1/Time']     = 1 / (X['Time_h'] + 1)
        X['Conc×Time']  = X['Concentration_ppm'] * X['Time_h']

        X_in = X[feature_names]
        X_sc = scaler.transform(X_in)
        rate = model.predict(X_sc)[0]

        status, color, rec = get_status_and_recommendation(rate, conc, time)

        # ---- pretty output ----
        print(f"{'Prediction Results':^50}")
        print("="*50)
        print(f"Inhibitor Dose : {conc:>8.1f} ppm")
        print(f"Exposure Time  : {time:>8.1f} hours")
        print("-"*50)
        print(f"Corrosion Rate : {rate:>8.4f} mm/year")
        print("="*50)
        color_code = {"green": "32", "orange": "33", "red": "31"}[color]
        print(f"Status         : \033[1;{color_code}m{status}\033[0m")
        print(f"Recommendation : {rec}")

# -------------------------------------------------
# 5. Bind ONLY on button click
# -------------------------------------------------
predict_btn.on_click(run_prediction)

# Optional: press Enter in a field → same as clicking button
def on_enter(change):
    if change['name'] == 'value' and change['owner'].value != change['old']:
        run_prediction(None)

conc_input.observe(on_enter, names='value')
time_input.observe(on_enter, names='value')

# -------------------------------------------------
# 6. Display
# -------------------------------------------------
print("CORROSION RATE PREDICTOR")
print("="*60)
display(conc_input, time_input, predict_btn, out)
````
