## **2. Experimental Method – Weight Loss Technique**

The **weight loss method** is the most reliable and industry-accepted technique for measuring corrosion rate. It involves:
1. Preparing mild steel coupons of known initial mass.
2. Immersing coupons in **corrosive medium (e.g., 1M HCl)** with and without inhibitor at controlled temperatures: **40°C, 50°C, 60°C, 70°C, 80°C**.
3. Retrieving coupons at fixed time intervals.
4. Cleaning, drying, and re-weighing to determine **mass loss**.
5. Calculating **corrosion rate** and **inhibitor efficiency**.

| :--- | :--- | :--- |
| **`Inhibition_efficiency_percent`** | % | The efficiency of the corrosion inhibitor is calculated as:  \n $\eta (\%)= \frac{W_0 - W_i}{W_0} \times 100$  \n Where:  \n - $W_0$ = Weight loss of the blank (uninhibited specimen).  \n - $W_i$ = Weight loss with inhibitor. |

## Step 3: Validating My Calculations, Exploring Summary Statistics + **Null Check**

**Goal**:  
Make sure **every number is correct** and **no data is missing** before any EDA.

**What We Check**:

| Check | Why |
|------|-----|
| 1. `Weight_loss_g` = `Initial - Final` | Lab error? |

````python
# --- STEP 3: FULL VALIDATION + NULL CHECK ---
print("STEP 3: DATA VALIDATION + NULL CHECK")
print("="*70)

# --- 0. NULL CHECK (FIRST!) ---
print("0. NULL VALUES IN EACH COLUMN")
print("-"*50)
null_counts = df.isnull().sum()
if null_counts.sum() == 0:
    print("No null values found!")
else:
    print(null_counts[null_counts > 0])
    print("\nWARNING: Nulls detected! Fix before EDA.")

# --- 1. Verify Weight Loss (g) ---
df['Calc_loss_g'] = df['Initial_mass_g'] - df['Final_mass_g']
loss_g_match = np.allclose(df['Weight_loss_g'], df['Calc_loss_g'], atol=1e-6)
print(f"\n1. Weight_loss_g correct? → {loss_g_match}")

# --- 2. Verify Weight Loss (mg) ---
df['Calc_loss_mg'] = df['Calc_loss_g'] * 1000
loss_mg_match = np.allclose(df['Weight_loss_mg'], df['Calc_loss_mg'], atol=1e-4)
print(f"2. Weight_loss_mg correct? → {loss_mg_match}")

# --- 3. RECALCULATE INHIBITOR EFFICIENCY (PER TEMPERATURE) ---
df['Efficiency_Corrected'] = 0.0

for temp in df['Temp_C'].unique():
    blank_row = df[(df['Temp_C'] == temp) & (df['Concentration_ppm'] == 0)]
    if blank_row.empty:
        print(f"   WARNING: No blank (0 ppm) at {temp}°C!")
        continue
    blank_loss = blank_row['Weight_loss_mg'].iloc[0]
    mask = df['Temp_C'] == temp
    df.loc[mask, 'Efficiency_Corrected'] = (1 - df.loc[mask, 'Weight_loss_mg'] / blank_loss) * 100

# Compare
eff_diff = (df['Inhibitor_efficiency'] - df['Efficiency_Corrected']).abs()
max_diff = eff_diff.max()
print(f"3. Efficiency recalculated. Max difference: {max_diff:.6f}%")

if max_diff > 1e-3:
    print("   → Replacing original with corrected values")
    df['Inhibitor_efficiency'] = df['Efficiency_Corrected']

# Clean up
df = df.drop(columns=['Efficiency_Corrected', 'Calc_loss_g', 'Calc_loss_mg'], errors='ignore')
````

### Q2: Is **Blank Corrosion Rate** **Exponential** with Temperature?

**Why This Question?**  
- Without inhibitor, corrosion should follow **Arrhenius law**:  
  → Rate = A × exp(−Ea / RT)  
- We want to know:  
  → Is the **rate increase exponential**?  
  → Or just **linear**?  
  → This tells us **how dangerous** high temperatures are **without protection**

**What We’ll Do**  
1. **Print exact blank corrosion rates** (mm/y) at each temperature  
2. **Plot rate vs temperature**  
3. **Add exponential fit** to test Arrhenius behavior  
4. **Show R²** — how well exponential fits

**Next**: Run **Code Cell Q2** → prints numbers → then say `plot q2`

````python
# --- Q2: PRINT BLANK CORROSION RATES ---
print("Q2: BLANK (0 ppm) CORROSION RATE (mm/y) vs TEMPERATURE")
print("="*70)

# Extract blank data
q2_data = df[df['Concentration_ppm'] == 0][['Temp_C', 'Corrosion_rate_mm_y']].copy()
q2_data = q2_data.sort_values('Temp_C').reset_index(drop=True)

# Print clean table
print(q2_data.round(3).to_string(index=False))

# Save for plot and fit
blank_temps = q2_data['Temp_C'].values
blank_rates = q2_data['Corrosion_rate_mm_y'].values

print("\nData ready for Q2 plot and exponential fit.")
````

````python
# --- Q2: PLOT BLANK CORROSION RATE + EXPONENTIAL FIT ---

from scipy.optimize import curve_fit

# Define exponential function
def exp_func(T, A, B):
    return A * np.exp(B * T)

# Fit curve
popt, _ = curve_fit(exp_func, blank_temps, blank_rates, p0=[1e-5, 0.05])
A_fit, B_fit = popt
rates_fit = exp_func(blank_temps, A_fit, B_fit)

# Calculate R-squared
ss_res = np.sum((blank_rates - rates_fit) ** 2)
ss_tot = np.sum((blank_rates - np.mean(blank_rates)) ** 2)
r2 = 1 - (ss_res / ss_tot)

# Plot
plt.figure(figsize=(10, 6))
plt.scatter(blank_temps, blank_rates, color='red', s=100, zorder=5, label='Data (0 ppm)')
plt.plot(blank_temps, blank_rates, 'ro-', linewidth=2, markersize=8)
plt.plot(blank_temps, rates_fit, 'k--', linewidth=2,
         label=f'Exponential Fit (R² = {r2:.4f})')

plt.title('Q2: Blank Corrosion Rate vs Temperature\n(Exponential Behavior?)')
plt.xlabel('Temperature (degrees C)')
plt.ylabel('Corrosion Rate (mm/y)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Exponential fit: Rate = {A_fit:.2e} * exp({B_fit:.4f} * T)")
print(f"R-squared = {r2:.4f}")
````

### Key Insights — Blank (0 ppm) Corrosion Rate vs Temperature

- Corrosion rate increases exponentially with temperature, showing a strong positive thermal effect on corrosion behavior.  
- Between 40°C and 70°C, the corrosion rate rises sharply from 44.05 mm/y to 227.53 mm/y, confirming temperature as a major driver of corrosion kinetics.  
- The exponential model (`Rate = 12.2 * exp(0.038 * T)`) fits the data well with an R² value of 0.8774, indicating that about 88% of the variation in corrosion rate is explained by temperature.  
- The rate increase begins to plateau slightly between 70°C and 80°C (227.5 → 231.7 mm/y), suggesting a possible surface saturation or film formation effect at very high temperatures.  
- Overall, the blank corrosion rate data strongly support the conclusion that corrosion accelerates exponentially with temperature in the absence of inhibitors.

````python
# --- Q5: PLOT + FIND 50% CROSSING TEMPERATURE ---
plt.figure(figsize=(10, 6))

critical_temps = {}

for col in q4_data.columns:
    eff = q4_data[col].values
    temps = q4_data.index.values

    plt.plot(temps, eff, marker='o', linewidth=2, markersize=6, label=col)

    # Find where efficiency drops below 50%
    if eff[0] > 50 and eff[-1] < 50:
        # Interpolate
        for i in range(len(eff) - 1):
            if eff[i] >= 50 and eff[i+1] < 50:
                t1, t2 = temps[i], temps[i+1]
                e1, e2 = eff[i], eff[i+1]
                t_cross = t1 + (50 - e1) * (t2 - t1) / (e2 - e1)
                critical_temps[col] = round(t_cross, 1)
                break
    elif eff[0] <= 50:
        critical_temps[col] = "<40"
    else:
        critical_temps[col] = "Never"

plt.axhline(50, color='red', linestyle='--', linewidth=2, label='50% Threshold')
plt.title('Q5: When Does Efficiency Fall Below 50%?')
plt.xlabel('Temperature (degrees C)')
plt.ylabel('Inhibitor Efficiency (%)')
plt.legend(title='Dose', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, alpha=0.3)
plt.ylim(0, 100)
plt.tight_layout()
plt.show()

# Print critical temperatures
print("CRITICAL TEMPERATURES (Efficiency < 50%):")
for dose, temp in critical_temps.items():
    print(f"  {dose}: {temp} degrees C")
````

### Q10: Can **Efficiency vs Concentration** Be Modeled **Smoothly**?

**Why This Question?**  
- Engineers need **dose-response curves** for optimization  
- We want to know:  
  → Is **efficiency increase smooth** with dose?  
  → Can we **interpolate** between 50, 100, 150 ppm?

**What We’ll Do**  
1. **Print average efficiency (%)** per dose (across all temperatures)  
2. **Plot** with **smooth curve fit** (cubic spline)  
3. **Show R²** of fit

**Next**: Run **Code Cell Q10 Print** → then **Code Cell Q10 Plot**

````python
# --- Q10: PRINT AVERAGE EFFICIENCY PER DOSE ---
print("Q10: AVERAGE INHIBITOR EFFICIENCY (%) PER DOSE (ALL TEMPS)")
print("="*70)

# Group by dose, mean efficiency (exclude blank)
q10_data = df[df['Concentration_ppm'] > 0].groupby('Concentration_ppm')['Inhibitor_efficiency'].mean()
q10_data = q10_data[[50, 100, 150, 200, 250]]
q10_data.index = [f"{int(x)} ppm" for x in q10_data.index]

print(q10_data.round(2).to_string())
````

````python
# --- Q10: PLOT EFFICIENCY vs CONCENTRATION WITH SMOOTH FIT ---
from scipy.interpolate import CubicSpline

# Extract data
concs = q10_data.index.str.replace(' ppm', '').astype(int).values
effs = q10_data.values

# Fit cubic spline
cs = CubicSpline(concs, effs)
x_fit = np.linspace(50, 250, 100)
y_fit = cs(x_fit)

# R-squared
y_pred = cs(concs)
ss_res = np.sum((effs - y_pred)**2)
ss_tot = np.sum((effs - np.mean(effs))**2)
r2 = 1 - (ss_res / ss_tot)

plt.figure(figsize=(10, 6))
plt.scatter(concs, effs, color='blue', s=100, zorder=5, label='Data')
plt.plot(x_fit, y_fit, 'darkblue', linewidth=3, label=f'Smooth Fit (R² = {r2:.4f})')
plt.title('Q10: Efficiency vs Inhibitor Concentration\n(Smooth Dose-Response)')
plt.xlabel('Concentration (ppm)')
plt.ylabel('Average Efficiency (%)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Smooth fit R-squared: {r2:.4f}")
print("Curve can be used to interpolate any dose between 50–250 ppm.")
````

### Key Insights — Average Inhibitor Efficiency per Dose

- Efficiency **increases steadily with dose**, from ~25% at 50 ppm to ~56% at 250 ppm.  
- The smooth fit with **R² = 1.0** confirms a nearly perfect, predictable relationship.  
- Implication: Efficiency vs. concentration can be **interpolated reliably**, allowing engineers to estimate performance for intermediate doses and optimize inhibitor usage.

## Feature Engineering: Build Smart Inputs for Prediction

**Goal**:  
Turn raw `Temp_C` + `Concentration_ppm` into **powerful features**  
Predict: `Corrosion_rate_mm_y` (and `Inhibitor_efficiency`)

**Why?**  
- Corrosion is **non-linear** with temperature  
- Inhibitor **desorbs exponentially** at high T  
- We need **physics-informed features**

---

### Features We’ll Create

| Feature | Formula | Why |
|-------|--------|-----|
| `1/T` | `1 / (Temp_C + 273.15)` | Arrhenius law |
| `log(Conc+1)` | `np.log1p(Concentration_ppm)` | Dose-response |
| `Temp²` | `Temp_C ** 2` | Non-linear heat |
| `Temp × Conc` | `Temp_C * Concentration_ppm` | Interaction |
| `High_Temp` | `Temp_C >= 70` | Critical zone |
| `Efficiency` | Already calculated | Target or feature |

---

**Next**: Run **Code Cell FE Print** → **Code Cell FE Plot**

````python
# --- FEATURE ENGINEERING: PRINT NEW FEATURES ---
print("FEATURE ENGINEERING: NEW INPUTS FOR MODEL")
print("="*65)

# Copy df
fe_df = df.copy()

# 1. Inverse temperature (Kelvin)
fe_df['Inv_T'] = 1 / (fe_df['Temp_C'] + 273.15)

# 2. Log concentration
fe_df['Log_Conc'] = np.log1p(fe_df['Concentration_ppm'])

# 3. Temp squared
fe_df['Temp_sq'] = fe_df['Temp_C'] ** 2

# 4. Interaction
fe_df['Temp_x_Conc'] = fe_df['Temp_C'] * fe_df['Concentration_ppm']

# 5. High temp flag
fe_df['High_Temp'] = (fe_df['Temp_C'] >= 70).astype(int)

# Select key columns
print_cols = ['Temp_C', 'Concentration_ppm', 'Inv_T', 'Log_Conc', 'Temp_sq', 'Temp_x_Conc', 'High_Temp', 'Corrosion_rate_mm_y']
fe_df[print_cols].round(4).head(10)
````

````python
# --- PLOT: FEATURE vs CORROSION RATE ---
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.ravel()

features = ['Temp_C', 'Concentration_ppm', 'Inv_T', 'Log_Conc', 'Temp_x_Conc', 'High_Temp']
titles = ['Temperature', 'Concentration', '1/T (Arrhenius)', 'Log(Conc+1)', 'Temp × Conc', 'High Temp (>=70)']

for i, (feat, title) in enumerate(zip(features, titles)):
    axes[i].scatter(fe_df[feat], fe_df['Corrosion_rate_mm_y'], alpha=0.7, s=50)
    axes[i].set_xlabel(title)
    axes[i].set_ylabel('Corrosion Rate (mm/y)')
    axes[i].grid(True, alpha=0.3)

plt.suptitle('Feature Engineering: Input vs Corrosion Rate', fontsize=16, y=0.98)
plt.tight_layout()
plt.show()

print("FEATURES READY: Strong patterns in 1/T and Temp×Conc")
````

## Predictive Modeling: Predict Corrosion Rate from Temp + Conc

**Features**: From **Feature Engineering** (Temp, Conc, 1/T, Log_Conc, Temp×Conc, High_Temp)

### Models We Will Use (Your List)

| Model | Why | Overfitting Guard |
|------|-----|-------------------|
| **Multiple Linear Regression** | Interpretable, baseline | None needed |
| **Polynomial Regression (Degree 2)** | Captures curves | **Limited to degree 2** |
| **Ridge Regression** | Handles collinearity | **α tuned via CV** |
| **Random Forest** | Complex interactions | **Shallow trees, min_samples_leaf=3** |

**All models use 5-fold CV**  
**Train on 24 rows, test on 6**  
**Final pick**: **Lowest MAE + stable**

**Next**: Run **Code Cell Model 1** → then **Code Cell Model 2** → ...  
We go **one model at a time**

````python
# --- MODEL 1: MULTIPLE LINEAR REGRESSION ---
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score, KFold
from sklearn.metrics import mean_absolute_error, r2_score

print("MODEL 1: MULTIPLE LINEAR REGRESSION")
print("="*60)

# Features from FE
X = fe_df[['Temp_C', 'Concentration_ppm', 'Inv_T', 'Log_Conc', 'Temp_sq', 'Temp_x_Conc', 'High_Temp']]
y = fe_df['Corrosion_rate_mm_y']

# 5-fold CV
kf = KFold(n_splits=5, shuffle=True, random_state=42)
mae_scores = -cross_val_score(LinearRegression(), X, y, cv=kf, scoring='neg_mean_absolute_error')
r2_scores = cross_val_score(LinearRegression(), X, y, cv=kf, scoring='r2')

print(f"MAE (CV): {mae_scores.mean():.2f} ± {mae_scores.std():.2f} mm/y")
print(f"R²  (CV): {r2_scores.mean():.3f} ± {r2_scores.std():.3f}")

# Fit final model
lr_model = LinearRegression().fit(X, y)
print("\nCoefficients:")
for name, coef in zip(X.columns, lr_model.coef_):
    print(f"  {name:12}: {coef:+8.3f}")
print(f"  Intercept   : {lr_model.intercept_:8.3f}")
````

````python
# --- PLOT: LR PREDICTIONS vs ACTUAL ---
y_pred = lr_model.predict(X)

plt.figure(figsize=(8, 6))
plt.scatter(y, y_pred, alpha=0.7, s=60)
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', lw=2)
plt.xlabel('Actual Corrosion Rate (mm/y)')
plt.ylabel('Predicted (mm/y)')
plt.title('Model 1: Linear Regression – Fit Quality')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

mae_final = mean_absolute_error(y, y_pred)
print(f"Final MAE (all data): {mae_final:.2f} mm/y")
````

# Key Insights – Multiple Linear Regression Model

- The model explains ~87% of the variance in corrosion rate (R² ≈ 0.874), with average prediction errors around 15.5 mm/y.  
- **Temperature** is the dominant driver of corrosion, both linearly and non-linearly (Temp_C negative, Temp_sq positive, High_Temp positive).  
- **Concentration** has a mild direct effect, but its transformed features (Log_Conc, Temp×Conc) capture most of its impact.  
- Interaction and non-linear terms are crucial: simple linear effects alone cannot explain the corrosion pattern.  
- The model predicts well overall, but the large transformed coefficients indicate feature scaling is critical for interpretation.  

**Summary:** Temperature drives corrosion, concentration reduces it, and non-linear effects are essential for accurate predictions.


````python
# --- MODEL 2: POLYNOMIAL REGRESSION (DEGREE 2) ---
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import KFold, cross_val_score
from sklearn.metrics import mean_absolute_error

print("MODEL 2: POLYNOMIAL REGRESSION (DEGREE 2)")
print("="*60)

# Features (same as before)
X = fe_df[['Temp_C', 'Concentration_ppm', 'Inv_T', 'Log_Conc', 'Temp_sq', 'Temp_x_Conc', 'High_Temp']]
y = fe_df['Corrosion_rate_mm_y']

# Degree 2 polynomial
poly_model = make_pipeline(PolynomialFeatures(degree=2, include_bias=False), LinearRegression())

# 5-fold CV
kf = KFold(n_splits=5, shuffle=True, random_state=42)
mae_scores = -cross_val_score(poly_model, X, y, cv=kf, scoring='neg_mean_absolute_error')
r2_scores = cross_val_score(poly_model, X, y, cv=kf, scoring='r2')

print(f"MAE (CV): {mae_scores.mean():.2f} ± {mae_scores.std():.2f} mm/y")
print(f"R²  (CV): {r2_scores.mean():.3f} ± {r2_scores.std():.3f}")

# Fit final model
poly_model.fit(X, y)
print("\nModel ready. Degree 2 captures curves safely.")
````

````python
# --- PLOT: POLY DEGREE 2 PREDICTIONS vs ACTUAL ---
y_pred = poly_model.predict(X)

plt.figure(figsize=(8, 6))
plt.scatter(y, y_pred, alpha=0.7, s=60, color='teal')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', lw=2)
plt.xlabel('Actual Corrosion Rate (mm/y)')
plt.ylabel('Predicted (mm/y)')
plt.title('Model 2: Polynomial Degree 2 – Fit Quality')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

mae_final = mean_absolute_error(y, y_pred)
print(f"Final MAE (all data): {mae_final:.2f} mm/y")
````

### Random Forest Features Explained

- **Inv_T**: Inverse of temperature (1 / T). Shows how corrosion changes at lower temperatures; high influence because corrosion often reacts non-linearly to temperature.  
- **Temp_C**: Actual temperature in °C. Corrosion generally increases with temperature, so this is an important driver.  
- **Temp_sq**: Temperature squared. Captures non-linear effects of temperature—like corrosion accelerating faster at high temperatures.  
- **High_Temp**: A flag/indicator for very high temperatures. Helps model recognize when temperature is extreme and corrosion spikes.  
- **Temp_x_Conc**: Interaction term between temperature and concentration. Shows combined effect, e.g., how corrosion changes if both temp and inhibitor change together. Minor impact here.  
- **Log_Conc**: Logarithm of inhibitor concentration. Used to model diminishing returns of concentration at higher doses. Low impact in this dataset.  
- **Concentration_ppm**: Raw inhibitor concentration. Surprisingly small effect here, likely because temperature dominates corrosion behavior.

````python
import joblib
import numpy as np

# Normalize R² and MAE to a common scale
r2_scores = comp_df['R² (CV)'].values
mae_scores = comp_df['MAE (all)'].values

# Invert MAE so that higher is better (like R²)
mae_inv = 1 / (mae_scores + 1e-6)  # add small number to avoid divide by zero

# Compute a combined score: higher R² and lower MAE
combined_score = r2_scores * 0.6 + mae_inv * 0.4  # weights can be adjusted

best_index = np.argmax(combined_score)
best_name = comp_df.loc[best_index, 'Model']
best_mdl = models[best_index]

# Save the model
joblib.dump(best_mdl, 'final_corrosion_model.pkl')

print(f"BEST MODEL: {best_name}")
print(f"R² (CV): {comp_df.loc[best_index, 'R² (CV)']:.3f}")
print(f"MAE (all): {comp_df.loc[best_index, 'MAE (all)']:.3f}")
print("Model saved as 'final_corrosion_model.pkl'")
````

## Interactive GUI (Temp + Conc → Predict)**


````python
import joblib
import numpy as np
import ipywidgets as widgets
from IPython.display import display, clear_output
import matplotlib.pyplot as plt

# Load model
model_pipeline = joblib.load('final_corrosion_model.pkl')

# --- Input widgets ---
style = {'description_width': '180px'}
layout = widgets.Layout(width='300px')

temp_input = widgets.BoundedFloatText(
    value=50.0, min=40.0, max=80.0,
    description='Temperature (°C):', style=style, layout=layout
)
dose_input = widgets.BoundedFloatText(
    value=150.0, min=50.0, max=250.0,
    description='Dose (ppm):', style=style, layout=layout
)
predict_btn = widgets.Button(description='Predict Corrosion Rate', button_style='success')
out = widgets.Output()

# --- Prediction function ---
def run_prediction(_):
    with out:
        clear_output()
        Temp_C = temp_input.value
        Concentration_ppm = dose_input.value

        # Feature engineering (must match training)
        Inv_T = 1 / (Temp_C + 273.15)
        Log_Conc = np.log1p(Concentration_ppm)
        Temp_sq = Temp_C ** 2
        Temp_x_Conc = Temp_C * Concentration_ppm
        High_Temp = 1 if Temp_C >= 70 else 0

        X_input = np.array([[Temp_C, Concentration_ppm, Inv_T, Log_Conc, Temp_sq, Temp_x_Conc, High_Temp]])
        rate = model_pipeline.predict(X_input)[0]

        # Determine status and color
        if rate < 5:
            status = "SAFE"
            color = 'green'
        elif rate < 50:
            status = "MONITOR CLOSELY"
            color = 'orange'
        else:
            status = "DANGER – SHUTDOWN NOW"
            color = 'red'

        # Colored print
        color_codes = {"green": "32", "orange": "33", "red": "31"}
        print(f"\033[1;{color_codes[color]}mTEMPERATURE: {Temp_C}°C\nDOSE: {Concentration_ppm} ppm\nCORROSION RATE: {rate:.2f} mm/y\nSTATUS: {status}\033[0m")

        # Bar chart
        plt.figure(figsize=(6,4))
        plt.bar(['Corrosion Rate'], [rate], color=color)
        plt.axhline(5, color='green', linestyle='--', label='Safe Limit')
        plt.axhline(50, color='orange', linestyle='--', label='Monitor Limit')
        plt.ylim(0, max(250, rate*1.2))
        plt.ylabel('Corrosion Rate (mm/y)')
        plt.title('Corrosion Risk Assessment')
        plt.legend()
        plt.grid(True, alpha=0.3)
        plt.show()

# --- Bind button ---
predict_btn.on_click(run_prediction)

# --- Display ---
display(temp_input, dose_input, predict_btn, out)
````
